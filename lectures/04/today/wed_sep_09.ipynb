```python
class Creature:
    def __init__(self, name):
        self.name = name
        print("Creature " + self.name + " created.")
    def talk(self):
        return "My name is " + self.name + " and I am a " + type(self).__name__ + self.makeSound();
    def makeSound(self):
        return ""
```

```python
a = Creature("Leslie")
```

```python
a
```

```python
a.talk()
```

```python
class Dog(Creature):
    def __init__(self, name):
        super().__init__(name)
    def makeSound(self):
        return ": woof."
```

```python
b = Dog("Spot")
```

```python
b.talk()
```

```python
class Complex:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag
    def __str__(self):
        return str(self.real) + " + i * " + str(self.imag)
    def __add__(self, other):
        return Complex(self.real + other.real, self.imag + other.imag)
    def __mul__(self, other):
        return Complex(self.real*other.real - self.imag*other.imag, self.real*other.imag + self.imag*other.real)
```

```python
a = Complex(1, 1)
```

```python
print(a)
```

```python
b = Complex(-2, 2)
```

```python
print(b)
```

```python
print(a + b)
```

```python
print(a * b)
```

So, what this wants to say is that $(1+i) \cdot (-2 + 2i) = -4$.
